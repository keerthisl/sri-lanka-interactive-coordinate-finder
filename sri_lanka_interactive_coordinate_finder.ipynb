```python
# KLA e.g.4f map made with folium - Sri Lanka with coordinates display, ver number on display
# Project name: sri-lanka-interactive-coordinate-finder
# Project by Keerthi Abe
# Version: 1.0.0 (March 11, 2025)


# Install required packages
!pip install folium

import folium
from google.colab import files
from IPython.display import IFrame, display

# Version information
VERSION = "1.0.0"
VERSION_DATE = "March 11, 2025"

# Create a map centered at Sri Lanka coordinates
m = folium.Map(location=[7.9569, 80.7597], zoom_start=10)

# Add a marker at the central point
folium.Marker([7.9569, 80.7597], popup="Sri Lanka - Central Point").add_to(m)

# Add a click event to show coordinates
m.add_child(folium.LatLngPopup())

# Add version information as a title in the corner of the map
title_html = f'''
    <div style="position: fixed;
                bottom: 10px; right: 10px; width: 200px; height: 50px;
                border:2px solid grey; z-index:9999; font-size:12px;
                background-color: white; padding: 5px;
                opacity: 0.7;">
        <b>Sri Lanka Map</b><br>
        By Keerthi Abe | v{VERSION} ({VERSION_DATE})
    </div>
'''
m.get_root().html.add_child(folium.Element(title_html))

# Save the map to an HTML file
output_file = "sri_lanka_map.html"
m.save(output_file)

# Read the file to add your name and version
with open(output_file, "r", encoding="utf-8") as file:
    html_content = file.read()

# Insert your name and version as HTML comments after the DOCTYPE
version_comment = f"<!-- Project by Keerthi Abe | Version: {VERSION} ({VERSION_DATE}) -->"
html_content = html_content.replace("<!DOCTYPE html>", f"<!DOCTYPE html>\n{version_comment}")

# Write the modified content back to the file
with open(output_file, "w", encoding="utf-8") as file:
    file.write(html_content)

# Rename the file to include "find_coordinates" in the name
import os
new_filename = f"sri_lanka_map_find_coordinates_v{VERSION}.html"
os.rename(output_file, new_filename)

print(f"Map saved to {new_filename} with your name and version added.")

# Try to display the map in Colab
try:
    display(m)
    print("Map displayed in notebook.")
except Exception as e:
    print(f"Could not display map in notebook: {e}")

# Download the HTML file to your local machine
files.download(new_filename)
print(f"Download of {new_filename} initiated.")
```
